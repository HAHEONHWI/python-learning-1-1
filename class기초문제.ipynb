```python
#1
class Person:
    def __init__(self, name):
        self.name = name

    def who(self):
        print(self.name)

p1 = Person("홍길동")
p1.who()
p2 = Person("이순신")
p2.who
```

```python
#2
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age

a = Animal('강아지', 3)
print(f"이름: {a.name} 나이 : {a.age}")
```

```python
#3
class Monster:
    def __init__(self, name):
        self.name = name

    def say(self):
        print(f"나는 {self.name}")

m = Monster("드래곤") 
m.say()
```

```python
#4
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def say(self):
        print(self.name, self.score)

s1 = Student('철수', 80)
s2 = Student('영희', 90)
s3 = Student('민수', 70)

s1.say()
s2.say()
s3.say()
print("================")

stu = [s1, s2, s3]
for i in stu:
    i.say()
```

```python
#5

class Score:
    def __init__(self, score):
        self.score = score

    def is_pass(self):
        if self.score >= 60 :
            print('합격')
        else:
            print('불합격')

s1 = Score(75)
s2 = Score(40)
s1.is_pass()
s2.is_pass()
```

```python
#6
class Student:
    school = "소프트웨어고"
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def intro(self):
        print(f"{self.school} 학생 {self.name}, {self.age}살")

s1 = Student("민주", 17)
s2 = Student("지우", 18)

s1.intro()
s2.intro()
```

```python
#7
class Book:
    category = "도서"
    def __init__(self, title, price):
        self.title = title
        self.price = price
        
    def discount(self, rate):
        self.price *= 1-rate

    def show(self):
        print(f"{self.title}/{self.price}/{self.category}")

b1 = Book("파이썬 입문", 20000)
b1.discount(0.1)
b1.show()
```

```python
#8
class Character:
    game_name = "RPG 게임"
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def attack(self, damage):
        self.hp -= damage

    def status(self):
        print(f"{self.name} HP:{self.hp} ({self.game_name})")


c1 = Character('전사', 100)
c2 = Character("마법사", 80)
c1.attack(20)
c2.attack(30)
c1.status()
c2.status()
```

```python
class Student:
    school = "소마고"
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def intro(self):
        print(f"{self.school} 학생 {self.name}입니다")

    def add_score(self, point):
        self.score += point
        print(f"{self.name}의 점수가 {point}점 증가했습니다")

    def show_score(self):
        print(f"{self.name}의 현재 점수 : {self.score}")

s1 = Student("철수", 80)
s2 = Student("영희", 90)
s1.intro()
s2.intro()
s1.add_score(10)
s2.add_score(5)
s1.show_score()
s2.show_score()
```
