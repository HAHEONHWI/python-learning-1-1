# 리스트 복습

```python
nums = [10, 20, 30, 40, 50]
for i in nums:
    print(i, end=' ')

print()

for j in range(len(nums)):
    print(nums[j], end=' ')
```

```python
nums = [3, 5, 7, 9]
n = 0
for i in nums :
    n += i
print('합:',n)

print()
total = 0
for j in range(len(nums)):
    total += nums[j]
print('합:',total)
```

```python
nums = [12, 7, 9, 20, 33, 18]
zak = []
for i in nums:
    if i%2 == 0:
        zak.append(i)
for j in zak:
    print(j, end=' ')

print()

zak = []
for j in range(len(nums)):
    if nums[j]%2 == 0:
        zak.append(nums[j])
for k in zak:
    print(k, end=' ')
    
```

```python
nums = [45, 12, 78, 34, 23]
maxi = 0
for i in nums:
    if maxi < i:
        maxi = i
print('최댓값:', maxi)
```
