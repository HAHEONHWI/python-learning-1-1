# 리스트 복습

```python
nums = [10, 20, 30, 40, 50]
for i in nums:
    print(i, end=' ')
```

```python
nums = [3, 5, 7, 9]
n = 0
for i in nums :
    n += i
print('합:',n)
```

```python
nums = [12, 7, 9, 20, 33, 18]
zak = []
for i in nums:
    if i%2 == 0:
        zak.append(i)
for j in zak:
    print(j, end=' ')
```
