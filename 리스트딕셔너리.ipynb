```python
#4

scores = {}

while True:
    data = input()

    if data == 'q':
        break
    else:
        name, score = data.split()
        scores[name] = int(score)

print(scores)

for who, num in scores.items():
    if num >= 80:
        print(who, '합격')
    else:
        print(who, '불합격')
```

```python
#6
howMany = int(input())

student = {}
max_score = 0
max_name = None

for i in range(howMany):
    name, score = input().split()
    student[name] = int(score)

for name, score in student.items():
    if score > max_score:
        max_score = score
        max_name = name

print(max_name, max_score)
```

```python
# 7
text = input()
seen = set()
dup = set()

for i in text:
    if i in seen:
        dup.add(i)
    else:
        seen.add(i)
print(dup)
```

```python
sentence = input("문장 입력: ").split()

freq = {}

for i in sentence:
    i = i.lower()
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)

for i, j in freq.items():
    print(i, j)
```

```python
# 9
pw = input()
info = {"alpha":0, "digit":0, "special":0}

for i in range(len(pw)):
    if pw[i].isalpha():
        info["alpha"] += 1
    elif pw[i].isdigit():
        info["digit"] += 1
    else:
        info["special"] += 1
    
print(info)
```

```python
key = list(input().split())
val = list(input().split())
dict = {}

for i in range(len(key)):
    dict[key[i]] = int(val[i])

print(dict)
```

```python
key_ = input().split()
val_ = list(map(int, input().split()))

dict1 = {}

for i, j in zip(key_, val_):
    dict1[i] = j

print(dict1)

person = {}

for i, j in zip(key_, val_):
    person[i] = j

print(person)
```

```python
# 11
fruits = ['사과', '바나나', '포도']
prices = [2500, 1500, 5000]
fruit_box = {}

for i, j in zip(fruits, prices):
    fruit_box[i] = j


print(fruit_box)

# 12

for f, p in fruit_box.items():
    print(f"{f}의 가격은 {p}원 입니다")

#13

sales_box = {}
for f, p in fruit_box.items():
    dsc_p = int(p*0.9)
    sales_box[f] = dsc_p

print(sales_box)
```

```python
# 14

names = ['민수', '영희', "철수"]
math = [80, 90, 70]
eng = [90, 85, 95]

for n, m, e in zip(names, math, eng):
    avg = (m+e)/2
    print(f"{n}의 평균점수 : {avg}")
```
