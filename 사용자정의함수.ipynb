# 7.사용자 정의 함수
> 학습지 내용

```python
def lst_create():
    lst = list(map(int, input().split()))
    print(f"리스트 생성 {lst}")
    return lst

def lst_cal(lst):
    total = sum(lst)
    avg = total / len(lst)
    return total, avg

def lst_append(lst):
    while True:
        data = input()
        if data == 'q':
            print(lst)
            return lst
        lst.append(int(data))

def lst_print(lst, result):
    print(f"점수 : {lst}")
    print(f"합계 : {result[0]}, 평균 : {result[1]}")

lst = lst_create()
lst = lst_append(lst)
result = lst_cal(lst)
lst_print(lst, result)
```
