```python
#클래스 정의
class Monster:
  def __init__(self, name, age):
    self.name = name #인스턴스 변수
    self.age = age
  def say(self):
    print(f'나는 {self.name} {self.age}살')
  # 객체 생성, 사용
shark = Monster('상어', 7)
wolf = Monster('늑대', 3)

shark.say()
wolf.say()
```

```python
class Dog:
    def __init__(self, name, age, kind):
        self.name = name
        self.age = age
        self.kind = kind
    def bark(self):
        print(f"{self.name}가 멍멍거린다~")
    def sit(self):
        print(f"{self.age}살 {self.kind}는 앉았다~")

my_dog = Dog("망고", 3, "푸들")
my_dog.bark()
your_dog = Dog("당근", 3, "진도")
your_dog.sit()
```

```python
class Cat:
    species = "고양이" #클래스 변수
    def __init__(self, name, color):
        self.name = name
        self.color = color
    def introduce(self):
        print(f"안녕 나는 {self.color}색 {self.species} {self.name}야!")

cat1 = Cat("나비", "흰")
cat2 = Cat("까미", "검정")
cat1.introduce()
cat2.introduce()
print(Cat.species)
```

```python
#상속과 오버라이딩
class Animal:
    def speak(self):
        print('동물이 소리를 냅니다');

class Dog(Animal):
    def speak(self):
        print('멍멍')

dog = Dog()
dog.speak()
```
