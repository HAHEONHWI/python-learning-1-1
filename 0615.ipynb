```python
#1
class Notification:
    platform = "공통 알림 시스템"
    def __init__(self, user):
        self.user = user
    def send(self, message):
        print(f"{self.user}에게 알림: {message}")

class EmailNotification(Notification):
    method = "이메일"
    def send(self,message):
        print(f"{self.user}에게 알림 : {message}")

class SMSNotification(Notification):
    method = '문자'
    def send(self,message):
        print(f"{self.user}에게 알림 : {message}")

class AppNotification(Notification):
    method = '앱푸시'
    def send(self,message):
        print(f"{self.user}에게 알림 : {message}")

email = EmailNotification("철수")
sms = SMSNotification('영희')
app = AppNotification('민수')

```

```python
#2
class Notification:
    platform = "공통 알림 시스템"
    def __init__(self, user):
        self.user = user
    def send(self, message):
        print(f"{self.user}에게 알림: {message}")

class EmailNotification(Notification):
    method = "이메일"
    def send(self,message):
        print(f"[이메일] {self.user}에게 알림 : {message}")

class SMSNotification(Notification):
    method = '문자'
    def send(self,message):
        print(f"[문자] {self.user}에게 알림 : {message}")

class AppNotification(Notification):
    method = '앱푸시'
    def send(self,message):
        print(f"[앱푸시] {self.user}에게 알림 : {message}")

email = EmailNotification("철수")
sms = SMSNotification('영희')
app = AppNotification('민수')


#(2)
print(email.user)
print(sms.user)
print(app.user)
print()
#(3)
print(Notification.platform)
print()
#4
print(email.platform)
print(sms.platform)
print(app.platform)
print()

#5
print(EmailNotification.method)
print(SMSNotification.method)
print(AppNotification.method)
print()

#6
print(email.method)
print(sms.method)
print(app.method)
print()

#7
email.send("회원가입 완료")
sms.send("인증번호 : 1234")
app.send('새 메시지가 도착했습니다')
print()

#8
Notification("관리자").send( "시스템 점검 안내")
print()

#9
email.user = '지민'
print(email.user)
email.send("비밀번호 변경")
print()

#10
notices = [email, sms, app]
for i in notices:
    i.send("공지사항이 있습니다")
print()
#11

#12
print(type(email))
print(type(sms))
print(type(app))
print()
```
