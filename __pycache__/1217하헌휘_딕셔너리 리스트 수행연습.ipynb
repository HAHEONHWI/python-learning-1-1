```python
#1
word = input()

if word == word[::-1]:
    print("팰린드롬")
else:
    print("팰린드롬이 아님.")
```

```python
#1 다른방식
word = input()
is_pal = True
for i in range(len(word)//2):
    if word[i] != word[-1-i]:
        is_pal = False
        break
if is_pal:
    print("팰린드롬")
else:
    print("팰린드롬 아님")
```

```python
nums = [10, 15, 13, 20]
for i in range(1, len(nums)):
    print(nums[i] - nums[i-1], end=' ')
```

```python
nums = [5, 7, 3, 8, 10]
for i in range(1, len(nums)):
    if nums[i] > nums[i-1]:
        print(nums[i], end=' ')
```
